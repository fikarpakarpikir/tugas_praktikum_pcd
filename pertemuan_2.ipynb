```python
img = cv2.imread("4-foto_profil.png")
print("Gambar Asli")
cv2_imshow(img)
```

```python
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("Gambar Gray")
cv2_imshow(img_gray)
```

```python
nilai_bg = 75

img_bright = cv2.add(img, nilai_bg)
print("Gambar Bright")
cv2_imshow(img_bright)
```

```python
img_terang = cv2.equalizeHist(img_gray)
print("Gambar Terang")
cv2_imshow(img_terang)
```
